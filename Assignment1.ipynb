#### 5.3

```python
def isPrime(n):
    if n<=2:
        return True
    for i in range(2,int(n**0.5) + 1):
        if n%i == 0:
            return False
    return True
num = int(input("Enter number:"))
sum = 0
for i in range(1,num+1):
    if isPrime(i):
        sum += i
print("Sum of primes till",num, "is",sum)
```

#### 6.1

```python
def addOdd(n):
    sum = 0
    for i in range(1,n+1):
        if i%2 == 1:
            sum += i
    return sum
n = int(input("Enter the number:"))
print("Sum of all odd numbers till",n,"is",addOdd(n))
```
