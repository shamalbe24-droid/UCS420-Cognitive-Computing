## Q.1
Write a program to create a NumPy 1-D array with 5 elements and perform:
- (a) Addition of 2 to all the elements
- (b) Multiply 3 with all the elements
- (c) Divide every element by 2

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])

print("Original array:", arr)
print("After adding 2:", arr + 2)
print("After multiplying by 3:", arr * 3)
print("After dividing by 2:", arr / 2)
```

## Q.2 (b)
Find the most frequent value and its index/indices in the following arrays.

### i.
`x = np.array([1,2,3,4,5,1,2,1,1,1])`

```python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
max_count = counts.max()
most_frequent_x = values[counts == max_count]

print("Array x:", x)
print("Most frequent value(s):", most_frequent_x)
print("Frequency:", max_count)

for value in most_frequent_x:
    print(f"Indices of {value}:", np.where(x == value)[0])
```

### ii.
`y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])`

```python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
max_count = counts.max()
most_frequent_y = values[counts == max_count]

print("Array y:", y)
print("Most frequent value(s):", most_frequent_y)
print("Frequency:", max_count)

for value in most_frequent_y:
    print(f"Indices of {value}:", np.where(y == value)[0])
```

## Q.3
For the given 2-D array:

`arr = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])`

Access:
- (a) 1st row, 2nd column
- (b) 3rd row, 1st column

NumPy uses **zero-based indexing**, so the corresponding indices are `[0, 1]` and `[2, 0]`.

```python
arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print("Array:")
print(arr)

print("1st row, 2nd column:", arr[0, 1])
print("3rd row, 1st column:", arr[2, 0])
```

## Q.4
Create a 1-D NumPy array named `Sarwangini` with evenly spaced 25 numbers from 10 to 100 using `linspace()`.

Print:
- dimensions of the array
- shape
- total elements
- data type of each element
- total number of bytes consumed

Then find the transpose using `reshape()`. Also check whether the same can be done using the `T` attribute.

```python
Sarwangini = np.linspace(10, 100, 25)

print("Sarwangini:")
print(Sarwangini)
print("\nDimensions (ndim):", Sarwangini.ndim)
print("Shape:", Sarwangini.shape)
print("Total elements (size):", Sarwangini.size)
print("Data type:", Sarwangini.dtype)
print("Total bytes consumed (nbytes):", Sarwangini.nbytes)

column_array = Sarwangini.reshape(25, 1)
transpose_using_reshape = column_array.reshape(1, 25)

print("\nTranspose using reshape():")
print(transpose_using_reshape)

transpose_using_T_direct = Sarwangini.T
print("\nDirect Sarwangini.T:")
print(transpose_using_T_direct)
print("Shape of Sarwangini.T:", transpose_using_T_direct.shape)

transpose_using_T = column_array.T
print("\nTranspose using T after reshape to 2-D:")
print(transpose_using_T)
print("Shape:", transpose_using_T.shape)
```

### Q.4 conclusion
A 1-D NumPy array has no separate row/column orientation, so `Sarwangini.T` alone does **not** change its shape. To obtain a row/column transpose, first make it 2-D with `reshape()` and then use `.T`.

## Q.5
Create a 2-D array of three rows and four columns named `ucs420_Sarwangini` using:

`10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35`

Compute:
- mean
- median
- maximum
- minimum
- unique elements

Then:
1. Reshape it to four rows and three columns and name it `reshaped_ucs420_Sarwangini`.
2. Resize it to two rows and three columns and name it `resized_ucs420_Sarwangini`.

```python
ucs420_Sarwangini = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original array (ucs420_Sarwangini):")
print(ucs420_Sarwangini)

print("\nMean:", np.mean(ucs420_Sarwangini))
print("Median:", np.median(ucs420_Sarwangini))
print("Maximum:", np.max(ucs420_Sarwangini))
print("Minimum:", np.min(ucs420_Sarwangini))
print("Unique elements:", np.unique(ucs420_Sarwangini))

reshaped_ucs420_Sarwangini = ucs420_Sarwangini.reshape(4, 3)

print("\nReshaped array (reshaped_ucs420_Sarwangini):")
print(reshaped_ucs420_Sarwangini)

resized_ucs420_Sarwangini = np.resize(ucs420_Sarwangini, (2, 3))

print("\nResized array (resized_ucs420_Sarwangini):")
print(resized_ucs420_Sarwangini)
```
