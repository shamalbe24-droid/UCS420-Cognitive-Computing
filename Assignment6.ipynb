## Q1. Sensor readings

Given the temperature readings and an average measurement error of +2°C, perform the operations using vectorization and Boolean indexing.

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print("a. Readings after adding 2°C:", corrected_temperature)

fahrenheit = corrected_temperature * 9 / 5 + 32
print("b. Temperatures in Fahrenheit:", fahrenheit)

readings_above_32 = corrected_temperature[corrected_temperature > 32]
print("c. Readings greater than 32°C:", readings_above_32)

count_above_32 = np.sum(corrected_temperature > 32)
print("d. Number of readings greater than 32°C:", count_above_32)
```

**e. Explanation:** Vectorization applies an operation to an entire NumPy array without writing an explicit loop. Boolean indexing selects elements that satisfy a condition. These approaches make code shorter and often faster because NumPy performs the operations in optimized array routines.

## Q2. Daily steps

The matrix contains the daily step counts of four users over three days.

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("a. Total steps recorded:", np.sum(steps))
print("b. Mean number of steps:", np.mean(steps))
print("c. Maximum steps:", np.max(steps))
print("   Minimum steps:", np.min(steps))
print("d. Total steps for each day (axis=0):", np.sum(steps, axis=0))
print("e. Total steps for each user (axis=1):", np.sum(steps, axis=1))

max_position = np.unravel_index(np.argmax(steps), steps.shape)
print("f. Maximum step count position (user index, day index):", max_position)
print("   Maximum step count:", steps[max_position])
```

## Q4. Predicting an assistance score

Each row of `X` represents a user. The columns represent sleep hours, activity level, and stress level. The coefficient vector is calculated using the Ordinary Least Squares equation given in the assignment.

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("a. Shape of X:", X.shape)
print("   Dimensions of X:", X.ndim)

X_transpose = X.T
print("\nb. X.T:")
print(X_transpose)
print("The transpose swaps rows and columns, so each feature becomes a row.")

XTX = X_transpose @ X
print("\nc. X.T @ X:")
print(XTX)

print("\nd. Inverse of X.T @ X:")
XTX_inverse = np.linalg.inv(XTX)
print(XTX_inverse)

beta = XTX_inverse @ X_transpose @ y
print("\ne. OLS coefficients (beta):")
print(beta)

print("\nf. Coefficient interpretation:")
print("Sleep-hours coefficient:", beta[0])
print("Activity-level coefficient:", beta[1])
print("Stress-level coefficient:", beta[2])
print("Each coefficient represents the model's change in predicted score for a one-unit increase in that feature, holding the other features constant.")

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("\ng. New user's predicted assistance score:", predicted_score)
```
